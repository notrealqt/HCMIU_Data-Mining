```python
import pandas as pd
data=pd.read_csv("user_positive_games.csv")
data.head()
```

```python
data.info()
```

```python
import ast

data['app_id'] = data['app_id'].apply(ast.literal_eval)  # Convert to lists
data = data[data['app_id'].apply(lambda x: len(x) > 1)]  # Filter
data.info()
```

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
transactions = data['app_id'].tolist() # Convert the 'app_id' column to a list of transactions

te = TransactionEncoder()
te_ary = te.fit_transform(transactions)
df = pd.DataFrame(te_ary, columns=te.columns_)
df.head()
```

```python
frequent_itemsets = apriori(df, min_support=0.001, use_colnames=True) # Lowered min_support
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
```

```python
print(rules)
```

```python
for index, row in rules.iterrows():
    antecedents = list(row['antecedents'])
    consequents = list(row['consequents'])

    print(f"{antecedents[0]} > {consequents[0]}")
```
