```python
import pandas as pd
data=pd.read_csv("user_positive_games.csv")
data.head()
```

```python
data.info()
```

```python
import ast

data['app_id'] = data['app_id'].apply(ast.literal_eval)  # Convert to lists
data = data[data['app_id'].apply(lambda x: len(x) > 3)]  # Filter
data.info()
```

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
transactions = data['app_id'].tolist() # Convert the 'app_id' column to a list of transactions

te = TransactionEncoder()
te_ary = te.fit_transform(transactions)
df = pd.DataFrame(te_ary, columns=te.columns_)
df.head()
```

```python
df.head()
```

```python
frequent_itemsets = apriori(df, min_support=0.004, use_colnames=True)

rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
```

```python
print(rules)
```

```python
# Sort rules by lift and confidence
best_rules = rules.sort_values(['lift', 'confidence'], ascending=[False, False])

# Print top 10 best rules with their metrics
print("Top 10 Best Association Rules:")
print("\n{:<15} {:<15} {:<10} {:<10} {:<10}".format(
    "Antecedent", "Consequent", "Lift", "Confidence", "Support"))
print("-" * 65)

for index, row in best_rules.head(10).iterrows():
    antecedents = list(row['antecedents'])
    consequents = list(row['consequents'])
    
    print("{:<15} {:<15} {:<10.2f} {:<10.2f} {:<10.4f}".format(
        str(antecedents[0]),
        str(consequents[0]),
        row['lift'],
        row['confidence'],
        row['support']
    ))
```
